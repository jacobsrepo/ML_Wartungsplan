```python
import pandas as pd

excel_path = PROJECT_ROOT / "data/raw/SAP_notintime_.xlsx"
book = pd.ExcelFile(excel_path, engine="openpyxl")
book.sheet_names
```

```python
snapshot = pd.read_excel(
    excel_path,
    sheet_name="Wartungsplan- Snapshot",
    engine="openpyxl",
)

print("Snapshot shape:", snapshot.shape)
display(snapshot.head())
display(
    snapshot["Aktuelle Wartungsstrategie (MPLA.STRAT)"]
    .replace(r"^\s*$", pd.NA, regex=True)
    .dropna()
    .value_counts()
    .to_frame("rows")
)
```

```python
orders = pd.read_excel(
    excel_path,
    sheet_name="Abfrage1",
    engine="openpyxl",
)

print("Order-history shape:", orders.shape)
display(orders.head())
display(orders.isna().mean().sort_values(ascending=False).head(20).to_frame("missing_share"))
```
